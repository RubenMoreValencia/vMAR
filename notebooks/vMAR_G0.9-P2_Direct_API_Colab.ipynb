```python
from vmar import Differential

modelo = Differential(field=lambda t, y: (-y[0],), initial_state=[1.0])
resultado = modelo.integrate(method='rk4', step=0.1, steps=10)
print('Estado final C23:', resultado.state)
print('Tiempo final:', resultado.times[-1])
```

```python
import math
referencia_r = 0.3678797744124984
exacta = math.exp(-1)
error_referencia = abs(resultado.state[0] - referencia_r)
error_analitico = abs(resultado.state[0] - exacta)
assert error_referencia < 1e-12
assert error_analitico < 1e-6
print('Paridad R–C23:', error_referencia)
print('Error frente a exacta:', error_analitico)
print('G0.9-P.2 DIRECT PYTHON API PASS')
```

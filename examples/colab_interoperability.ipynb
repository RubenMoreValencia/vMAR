```python
from vmar import Differential, Network, NetworkMechanism, ControlledNeuron, memory_ema, ConfirmedTrace
r = Differential(lambda t, y: (-y[0],), [1.]).integrate(steps=10)
print('RK4 state:', r.state)
n = Network((2, 4, 1))
x, y = [[0., 1.], [1., 0.]], [0., 1.]
print('loss before:', n.loss_gradient(x, y)[0])
n.step(x, y, step=.1)
print('loss after:', n.loss_gradient(x, y)[0])
print('memory:', memory_ema([0., 0.], [1., 2.], .8))
print('mode M1:', NetworkMechanism(n, x, y, mode='observe').step())
print('controlled:', ControlledNeuron([[1.]], [1.], [0., 0.]).step())
trace = ConfirmedTrace(2, 2)
trace.append(0, 0., 1., 0., [0., 0.])
trace.append(1, .1, .8, .1, [.1, .2])
print('confirmed:', trace.records)
```
